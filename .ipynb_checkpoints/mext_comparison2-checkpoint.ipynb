## Import Corpora and Extract Text

```python
import pdfplumber
import pandas as pd

# Define your corpus files and metadata
corpus_files = [
    {
        "path": "kaisetsu_2008_fla.pdf",
        "corpus": "2008",
        "type": "activity",
        "grade": "5-6"
    },
    {
        "path": "kaisetsu_2017_fla.pdf",
        "corpus": "2017",
        "type": "activity",
        "grade": "3-4"
    },
    {
        "path": "kaisetsu_2017_fls.pdf",
        "corpus": "2017",
        "type": "subject",
        "grade": "5-6"
    },
]

def extract_text(path):
    text = []
    with pdfplumber.open(path) as pdf:
        for i, page in enumerate(pdf.pages):
            t = page.extract_text()
            if t:
                text.append({"page": i+1, "text": t})
    return text

# Extract all three
results = {}
for doc in corpus_files:
    results[doc["corpus"] + "_" + doc["type"]] = extract_text(doc["path"])

# Print Page 1 of each file to confirm successful extraction
for key, pages in results.items():
    print(f"\n=== {key} — page 1 ===")
    print(pages[0]["text"][:500])
```

```python
#Build full data frame
# Section tagging based on known page ranges
section_map = {
    "2008_activity": {
        "sousetsu": range(1, 8),      # p.1-8
        "pedagogical": range(9, 29)   # p.9-29
    },
    "2017_activity": {
        "sousetsu": range(1, 9),     # p.1-9 
        "pedagogical": range(10, 54)  # p.10-54
    },
    "2017_subject": {
        "sousetsu": range(1, 9),     # p.1-9 
        "pedagogical": range(10, 81)  # p.10-81
    }
}

rows = []
for doc in corpus_files:
    key = doc["corpus"] + "_" + doc["type"]
    pages = results[key]
    for page_data in pages:
        p = page_data["page"]
        tag_map = section_map[key]
        if p in tag_map["sousetsu"]:
            section = "sousetsu"
        elif p in tag_map["pedagogical"]:
            section = "pedagogical"
        else:
            section = "other"
        rows.append({
            "corpus": doc["corpus"],
            "type": doc["type"],
            "grade": doc["grade"],
            "key": key,
            "page": p,
            "section": section,
            "text": page_data["text"]
        })

df = pd.DataFrame(rows)
print(df.groupby(["key", "section"])["page"].count())
```

## Tokenization

```python
from sudachipy import tokenizer
from sudachipy import dictionary

sudachi = dictionary.Dictionary().create()
mode = tokenizer.Tokenizer.SplitMode.C

def tokenize_text(text):
    tokens = []
    for token in sudachi.tokenize(text, mode):
        tokens.append({
            "surface": token.surface(),
            "lemma": token.dictionary_form(),
            "pos": token.part_of_speech()[0]
        })
    return tokens

# Apply to full dataframe
df["tokens"] = df["text"].apply(tokenize_text)

# Confirm
print(f"Tokenization complete. {len(df)} pages processed.")
print(f"Sample token count from first pedagogical page of 2008:")
sample = df[(df["key"] == "2008_activity") & (df["section"] == "pedagogical")].iloc[0]["tokens"]
print(f"{len(sample)} tokens")
```

```python
#gives the token counts per corpus per section, 
#and shows exactly how much analytical material is in each

# Filter to nouns and verbs only, using lemma form
def get_content_words(tokens):
    return [t["lemma"] for t in tokens if t["pos"] in ["名詞", "動詞"]]

df["content_words"] = df["tokens"].apply(get_content_words)

# Quick summary
for key in df["key"].unique():
    for section in ["sousetsu", "pedagogical"]:
        subset = df[(df["key"] == key) & (df["section"] == section)]
        total_words = sum(len(w) for w in subset["content_words"])
        print(f"{key} | {section} | {total_words} content words")
```

```python
# Form-related terms (19 terms)
form_terms = [
    "文法", "文構造", "表現", "語彙", "文型", "音声",
    "発音", "文字", "語順", "言語材料", "語句",
    "文", "文法的", "構造", "習得", "練習",
    "反復", "慣れ親しむ", "繰り返す"
]

# Meaning-related terms (19 terms)
meaning_terms = [
    "意味", "内容", "伝える", "理解", "コミュニケーション",
    "気持ち", "考え", "伝わる", "思い",
    "意図", "活用", "やり取り",
    "思考", "判断", "伝え合う", "目的",
    "場面", "工夫", "相手"
]
```

## Frequency Analysis

```python
# Frequency Analysis

from collections import Counter

def count_terms(content_words, term_list):
    counts = Counter()
    for word in content_words:
        if word in term_list:
            counts[word] += 1
    return counts

# Build summary dataframe
rows = []
for _, group in df.groupby(["key", "section"]):
    all_words = [w for words in group["content_words"] for w in words]
    total = len(all_words)
    
    for category, terms in [
        ("form", form_terms),
        ("meaning", meaning_terms)
    ]:
        counts = count_terms(all_words, terms)
        raw = sum(counts.values())
        per_1000 = round((raw / total) * 1000, 2) if total > 0 else 0
        rows.append({
            "corpus": group["corpus"].iloc[0],
            "type": group["type"].iloc[0],
            "section": group["section"].iloc[0],
            "category": category,
            "raw_count": raw,
            "total_words": total,
            "per_1000": per_1000
        })

freq_df = pd.DataFrame(rows)
print(freq_df.to_string(index=False))
```

```python
#Visualize frequency analysis results with a bar graph

import matplotlib.pyplot as plt
import japanize_matplotlib
import numpy as np

ped = freq_df[freq_df["section"] == "pedagogical"].copy()
ped["label"] = ped["corpus"] + "\n" + ped["type"]

categories = ["form", "meaning"]
labels = ped["label"].unique()
x = np.arange(len(labels))
width = 0.2

fig, ax = plt.subplots(figsize=(10, 6))
for i, cat in enumerate(categories):
    vals = [ped[(ped["label"] == l) & 
               (ped["category"] == cat)]["per_1000"].values[0] 
            for l in labels]
    ax.bar(x + i * width, vals, width, label=cat)

ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(labels)
ax.set_ylabel("Frequency per 1,000 content words")
ax.set_title("Term category frequency by corpus (pedagogical sections)")
ax.legend()
plt.tight_layout()
plt.savefig("freq_by_corpus.png", dpi=150)
plt.show()
```

## Collocation Analysis

```python
# Collocation Analysis

def get_collocates(df_subset, target_term, window=3):
    """Get all tokens within window positions of target term"""
    collocates = []
    for tokens in df_subset["tokens"]:
        surfaces = [t["lemma"] for t in tokens]
        for i, token in enumerate(surfaces):
            if token == target_term:
                left = max(0, i - window)
                right = min(len(surfaces), i + window + 1)
                context = surfaces[left:i] + surfaces[i+1:right]
                collocates.extend(context)
    return Counter(collocates)

# Define stopwords to filter trivial collocates
stopwords = set([
    "する", "ある", "いる", "なる", "こと", "もの", "ため",
    "など", "また", "その", "この", "これ", "それ", "よう",
    "ない", "れる", "られる", "て", "に", "を", "は", "が",
    "で", "の", "と", "も", "から", "まで", "より", "や",
    "か", "ば", "たり", "ながら", "おり", "おく", "いく",
    "くる", "みる", "しまう", "つける", "させる", "できる",
    "とも", "だり"
])

def get_filtered_collocates(df_subset, target_term, window=3, topn=10):
    raw = get_collocates(df_subset, target_term, window)
    filtered = {k: v for k, v in raw.items() 
                if k not in stopwords and k != target_term and len(k) > 1}
    return Counter(filtered).most_common(topn)

# Test on 慣れ親しむ across all three pedagogical corpora
target = "慣れ親しむ"
for key in ["2008_activity", "2017_activity", "2017_subject"]:
    subset = df[(df["key"] == key) & (df["section"] == "pedagogical")]
    results = get_filtered_collocates(subset, target, window=3, topn=10)
    print(f"\n=== {key} — collocates of {target} ===")
    for word, count in results:
        print(f"  {word}: {count}")
```

```python
# Define all target terms with their tokenized forms
target_terms = {
    # Single tokens
    "慣れ親しむ": ["慣れ親しむ"],
    "コミュニケーション": ["コミュニケーション"],
    "外国語": ["外国語"],
    "表現力": ["表現力"],
    # Bigrams
    "言語活動": ["言語", "活動"],
    "言語材料": ["言語", "材料"],
    "文構造": ["文", "構造"],
    "深い学び": ["深い", "学び"],
}

def get_collocates_universal(df_subset, token_list, window=3, topn=10):
    """Works for both single tokens and bigrams"""
    n = len(token_list)
    collocates = []
    
    for tokens in df_subset["tokens"]:
        lemmas = [t["lemma"] for t in tokens]
        for i in range(len(lemmas) - n + 1):
            if lemmas[i:i+n] == token_list:
                left = max(0, i - window)
                right = min(len(lemmas), i + n + window)
                context = lemmas[left:i] + lemmas[i+n:right]
                collocates.extend(context)
    
    filtered = Counter({
        k: v for k, v in Counter(collocates).items()
        if k not in stopwords 
        and k not in token_list
        and len(k) > 1
    })
    return filtered.most_common(topn)

# Run on 言語活動 across all three pedagogical corpora
for key in ["2008_activity", "2017_activity", "2017_subject"]:
    subset = df[(df["key"] == key) & (df["section"] == "pedagogical")]
    results = get_collocates_universal(
        subset, target_terms["言語活動"], window=3, topn=10
    )
    print(f"\n=== {key} — collocates of 言語活動 ===")
    for word, count in results:
        print(f"  {word}: {count}")
```

```python
# Full systematic collocation run
keys = ["2008_activity", "2017_activity", "2017_subject"]
sections = ["pedagogical", "sousetsu"]

for term_label, token_list in target_terms.items():
    print(f"\n{'='*60}")
    print(f"TARGET: {term_label}")
    print(f"{'='*60}")
    for key in keys:
        for section in sections:
            subset = df[(df["key"] == key) & (df["section"] == section)]
            results = get_collocates_universal(
                subset, token_list, window=3, topn=5
            )
            if results:
                print(f"\n  {key} | {section}")
                for word, count in results:
                    print(f"    {word}: {count}")
            else:
                print(f"\n  {key} | {section} — no occurrences")
```

## Keyness Analysis

```python
import numpy as np

def log_likelihood(a, b, total_a, total_b):
    """
    Calculate log-likelihood keyness score
    a = frequency in target corpus
    b = frequency in reference corpus
    total_a = total tokens in target corpus
    total_b = total tokens in reference corpus
    """
    if a == 0 and b == 0:
        return 0
    E1 = total_a * (a + b) / (total_a + total_b)
    E2 = total_b * (a + b) / (total_a + total_b)
    ll = 0
    if a > 0:
        ll += 2 * a * np.log(a / E1)
    if b > 0:
        ll += 2 * b * np.log(b / E2)
    # Positive = key in target, negative = key in reference
    if a / total_a < b / total_b:
        ll = -ll
    return round(ll, 2)

def get_corpus_freq(df, key, section="pedagogical"):
    """Get lemma frequencies for a corpus section"""
    subset = df[(df["key"] == key) & (df["section"] == section)]
    all_words = [w for words in subset["content_words"] for w in words]
    return Counter(all_words), len(all_words)

# Build frequency profiles for all three pedagogical sections
freq_2008, total_2008 = get_corpus_freq(df, "2008_activity")
freq_2017a, total_2017a = get_corpus_freq(df, "2017_activity")
freq_2017s, total_2017s = get_corpus_freq(df, "2017_subject")

print(f"2008 activity pedagogical: {total_2008} tokens")
print(f"2017 activity pedagogical: {total_2017a} tokens")
print(f"2017 subject pedagogical:  {total_2017s} tokens")
```

```python
def keyness_comparison(freq_target, total_target, freq_ref, total_ref, topn=20):
    """Return top key terms in target relative to reference"""
    all_terms = set(freq_target.keys()) | set(freq_ref.keys())
    scores = []
    for term in all_terms:
        a = freq_target.get(term, 0)
        b = freq_ref.get(term, 0)
        # Filter low frequency terms
        if a + b < 5:
            continue
        ll = log_likelihood(a, b, total_target, total_ref)
        scores.append((term, ll, a, b))
    scores.sort(key=lambda x: x[1], reverse=True)
    return scores[:topn], scores[-topn:]

# Comparison 1: 2008 vs 2017 subject (primary)
pos1, neg1 = keyness_comparison(freq_2008, total_2008, freq_2017s, total_2017s)
print("\n=== KEY IN 2008 vs 2017 subject ===")
for term, ll, a, b in pos1:
    print(f"  {term}: LL={ll} (2008={a}, 2017s={b})")
print("\n=== KEY IN 2017 subject vs 2008 ===")
for term, ll, a, b in neg1:
    print(f"  {term}: LL={ll} (2008={a}, 2017s={b})")

# Comparison 2: 2017 activity vs 2017 subject
pos2, neg2 = keyness_comparison(freq_2017a, total_2017a, freq_2017s, total_2017s)
print("\n=== KEY IN 2017 activity vs 2017 subject ===")
for term, ll, a, b in pos2:
    print(f"  {term}: LL={ll} (2017a={a}, 2017s={b})")
print("\n=== KEY IN 2017 subject vs 2017 activity ===")
for term, ll, a, b in neg2:
    print(f"  {term}: LL={ll} (2017a={a}, 2017s={b})")

# Comparison 3: 2008 vs 2017 activity
pos3, neg3 = keyness_comparison(freq_2008, total_2008, freq_2017a, total_2017a)
print("\n=== KEY IN 2008 vs 2017 activity ===")
for term, ll, a, b in pos3:
    print(f"  {term}: LL={ll} (2008={a}, 2017a={b})")
print("\n=== KEY IN 2017 activity vs 2008 ===")
for term, ll, a, b in neg3:
    print(f"  {term}: LL={ll} (2008={a}, 2017a={b})")
```

```python
# Save dataframe
df.drop(columns=["tokens"]).to_csv("kaisetsu_corpus.csv", index=False)
freq_df.to_csv("frequency_results.csv", index=False)

# Save keyness results
import json
keyness_results = {
    "2008_vs_2017subject": {"key_in_2008": pos1, "key_in_2017s": neg1},
    "2017activity_vs_2017subject": {"key_in_2017a": pos2, "key_in_2017s": neg2},
    "2008_vs_2017activity": {"key_in_2008": pos3, "key_in_2017a": neg3}
}
with open("keyness_results.json", "w", encoding="utf-8") as f:
    json.dump(keyness_results, f, ensure_ascii=False, indent=2)

print("Files saved.")
```
